## Математические методы

Вопрос о том, чем на самом деле являются точки, прямые и числа, не имеет смысла, поскольку предметом математики является структура этих объектов и их взаимосвязи. Основные математические концепции это непрерывность, движение и бесконечность. Бесконечное в математике связывается с некоторым процессом, не имеющим конца, и никогда не связывается с актуальной величиной.

Математические методы балансируют между строгостью логических рассуждений, творчеством и интуицией. 

Исторически первым направлением были **аксиоматико-дедуктивные методы**, берущие начало со времен Евклида. В нем числа определяются как точки на плоскости, а алгебраические свойства доказываются путем решения геометрических проблем, где каждый логический шаг можно изобразить визуально. Примером такого доказательства для теоремы Пифагора:

![Доказательство через равнодополняемость](https://habrastorage.org/r/w1560/webt/-0/hk/7x/-0hk7xxzoninl-f3-2tsjqvugu4.png)

Равенство $a^2+b^2=c^2$ доказывается через равенство площадей вписанных в квадрат фигур. Из рисунка видно, что слева и справа площади закрашенных фигур равны, так как в каждом из больших квадратов закрашено по 4 одинаковых прямоугольных треугольника. А это означает, что и незакрашенные (белые) площади слева и справа тоже равны. Замечаем, что в первом случае площадь незакрашенной фигуры равна $c^2$, а во втором — площадь незакрашенной области равна $a^2 + b^2$. Таким образом, $c^2 = a^2+b^2$.

Похожим образом мы можем доказать равенство $(a + b)^2 = a^2 + 2ab + b^2$:

В XVII веке Декарт и Ферма совершили алгебраическую революцию. Декарт предложил **формальную процедуру (алгоритм)**. Он ввел систему координат и символьные нотации, превратив геометрические фигуры в уравнения $(x,y)$ и доказав их взаимозаменяемость. 

> Декарт и Ферма показали, что любая кривая (геометрия) может быть описана уравнением (алгебра). 

В аналитической геометрии операции с символами происходят в отрыве от их геометрических значений, что, в частности, позволяло механизировать процесс решения задач. Для получения результата достаточно следовать формулам и правилам переноса из разных частей уравнений.


Конфликт между строгостью доказательства и интуитивностью методов достиг своего пика, когда Ньютон и Лейбниц ввели исчисление бесконечно малых. В Евклидовых методах все величины конечны и определены. В дифференциальных методах появились сущности $dx$ — величины, которые «больше нуля, но меньше любого мыслимого числа». Потребовалось еще 200 лет, чтобы под эти мистические методы подвели новый аксиоматический фундамент. 

## Арифметические операции

Арифметика это теория натуральных чисел. Для сложения и умножения на множестве натуральных чисел устанавливаются коммутативный и ассоциативный законы, а также дистрибутивный сложения относительно умножения. Это верно только для чисел - в химии мы не можем говорить о коммутативности "сложения" воды с серной кислотой. Неравенство $a$ относительно $b$ определяется через такое $с$, которое может быть добавлено, чтобы получить равенство этих чисел. Через это $c$ возможно определить операцию вычитания. 

> На множестве натуральных чисел вычитание тесно связано с упорядоченностью, поскольку $a - b$ определено, только если $a \geq b$, в силу отсутствия обратных элементов. Но можно определить вычитание и без порядка, например, если структура обладает групповыми свойствами (сложение ассоциативно, есть нейтральный элемент, у каждого элемента есть обратный). Тогда вычитание всегда определено $a - b = a + (-b)$, где $-b$ — обратный элемент. Пример таких групп - векторные пространства и комплексные числа

2. Напишите "тридцать" и "сто тридцать три" в системах с основаниями 7, 11, 12

![](images/div.png)

- $30_{10} = 42_7$
- $133_{10} = 250_7$
- $30_{10} = 28_{11}$
- $133_{10} = 111_{11}$
- $30_{10} = 26_{12}$
- $133_{10} = b1_{12}$

```python
def convert_to(number, base):
    if number == 0:
        return '0'
    digits = '0123456789abcdefghijklmnopqrstuvwxyz'
    result = ''
    while number > 0:
        result = digits[number % base] + result
        number //= base
    return result

convert_to(133, 11)
```
