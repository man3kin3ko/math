<h1>Теория чисел</h1>

Числа есть результат абстрагирования. Например, число 6 - результат абстрагирования всех возможных совокупностей, состоящих из шести предметов. Для детей числа остаются связанными с их материальными воплощениями, не выделяясь в отдельную концепцию.

<h2>Натуральные числа</h2> 

Исторически они определялись через **аксиомы Пеано**, который, в частности, обозначил 1 первым числом натуральной последовательности. С появлением теории множеств натуральные числа также определяются как $0=\varnothing, S(n)=n\cup\{n\}$, где $S(n)$ является функцией следования, ставящая в соответствие натуральному числу следующее за ним натуральное число.

> Вопрос включения нуля в натуральные числа является скорее политическим. Некоторые подчеркивают удобство такого использования, другие обращают внимания на физический смысл натуральных чисел как счетчика объектов, где понятие отсутствия объекта появилось значительно позже.

Поскольку натуральные числа рекурсивно определяются как переход от $n$ к $n+1$, это их фундаментальное свойство используется в основе метода **математической индукции** - если некое утверждение $А$ справедливо для единицы и доказано, что при переходе от $A_n$ к $A_n + 1$ оно остается верным, то оно верно для всех натуральных чисел.

Мы примем натуральные числа как данные вместе с двумя основными операциями сложения и умножения. Геометрический смысл этих операций легко показать на примере точек в ящичках (абак)

Математическую теорию натуральных чисел, основанную на действии нескольких законов, называют арифметикой. К этим законам относятся:
1) $a + b = b + a$ - коммутативный (переместительный) по сложению
2) $ab = ba$ - коммутативный по умножению
3) $a + (b + c) = (a + b) + c$ - ассоциативный (сочетательный) по сложению
4) $a(bc) = (ab)c$ - ассоциативный по умножению
5) $a(b + c) = ab + bc$ - дистрибутивный (распределительный)

<h2>Целые числа</h2> 

Введение *нотации* отрицательных чисел и нуля позволило неограниченно производить *обратную операцию* к сложению. Действительно, введем понятие неравенства $a < b$ и $b > a$, держа в уме геометрический смысл с ящичками. Тогда $a < b$ если ящик $b$ может быть получен из $a$ прибавлением $c$, или $b = a + c$. Тогда мы можем определить операцию вычитания $c = b - a$

<h2>Рациональные числа</h2> 

Натуральные и целые числа **дискретны**. Рациональные числа появились в результате необходимости измерять **непрерывные** величины. Это результат абстракции процесса измерения, при котором величина не делилась без остатка на единицу измерения, и этот остаток $m$ измерялся введением подъединицы $n$. Таким образом рациональное число - это число, представленное в виде дроби $\frac{m}{n},$ где $m,n \in \mathbb{N}$.

<h2>Иррациональные числа</h2>

Не смотря на то, что представление чисел в виде отрезков на координатной прямой уже некоторое время было в ходу, на прямой можно отложить такие точки, которые невозможно представить в виде дроби - например, диагональ квадрата с единичной стороной. Чтобы ввести взаимное соответствие между точками на прямой и числами, необходимо было ввести иррациональные числа

>“Irrational” means a number cannot be written as a ratio — a fraction

<h2>Комплексные числа</h2>

В отличии от действительных чисел, комплексные числа двумерны, поскольку их векторное пространство - комплексная плоскость - задается двумя базисными векторами, единицей и мнимой единицей




<h3>Представление чисел</h3>

Существующая **нумерация** (представление чисел с помощью знаков) явялется **позиционной**, в то время как римская запись являлась **аддититвной**. 

В числе, записываемом $abcd$, коэффициенты представляют *остатки от деления* на основание системы счисления.

В **каноническом разложении** число записывается как произведение простых сомножителей в некоей степени - $p_1^{\alpha_1}p_2^{\alpha_2}...p_n^{\alpha_n}$. Это разложение на множители возможно единственным образом - основная теорема арифметики
